```python
import numpy as np
```

```python
class Logistic_regression():

  #declare hyperparameters(learing rate and the epochs)
  def __init__(self,learning_rate,no_of_iterations):
    self.learning_rate=learning_rate
    self.no_of_iterations=no_of_iterations

#declare the features and output
  def fit(self,X,Y):
    #no. of features and datpoints in the input features(X) for the derivative formulas
    self.m,self.n=X.shape
    #inititate values of  weights and bias(model parameters)
    self.w=np.zeros()
    self.b=0
    self.X=X
    self.Y=Y
    #implement gradient descent for model optimization

    for i in range(no_of_iterations):
      self.update_weights


  def update_weights(self):
    #the Y hat formula (predicted Y)
    #substitute z with (w.x+b)
    Y_hat=1/(1+np.exp(-(self.X.dot(self.w)+self.b)))

    #find the derivatives for updating weights
    dw=(1/self.m)*np.dot(self.X.T,(Y_hat-self.Y))
    db=(1/self.m)*np.sum(Y_hat-self.Y)

    #update the weights and bias using gradient descent
    self.w=self.learning_rate*dw
    self.b=self.learning_rate*db

    #create the sigmoid equation and the decision boundary (0-1)

  def predict(self,X):

    Y_pred=1/(1+np.exp(-(self.X.dot(self.w)+self.b)))
    Y_pred=np.where(Y_pred>0.5,1,0)
    return Y_pred


```
